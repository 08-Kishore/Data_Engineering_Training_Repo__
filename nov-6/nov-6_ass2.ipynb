```python

registrations_df = spark.table("event_registrations_source")
events_df = spark.table("event_master")
registrations_df.printSchema()
events_df.printSchema()
```

```python
from pyspark.sql.functions import col

selected_df = registrations_df.select(
    "registration_id",
    "attendee_name",
    "tickets",
    "unit_price"
)

selected_df.show()
```

```python
from pyspark.sql.functions import col

filtered_df = registrations_df.filter(
    (col("status") == "Confirmed") &
    (col("tickets") > 2)
)

filtered_df.show()
```

```python
from pyspark.sql.functions import col

amount_df = registrations_df.withColumn(
    "total_amount",
    col("tickets") * col("unit_price")
)

amount_df.show()
```

```python
from pyspark.sql.functions import col, when

booking_df = (
    registrations_df
    .withColumn(
        "total_amount",
        col("tickets") * col("unit_price")
    )
    .withColumn(
        "booking_category",
        when(col("total_amount") >= 2000, "High")
        .otherwise("Regular")
    )
)

booking_df.show()
```

```python
from pyspark.sql.functions import col, sum

joined_df = registrations_df.alias("r").join(
    events_df.alias("e"),
    col("r.event_id") == col("e.event_id"),
    "inner"
)

revenue_df = (
    joined_df
    .filter(col("r.status") == "Confirmed")
    .groupBy("e.city")
    .agg(
        sum(
            col("r.tickets") * col("r.unit_price")
        ).alias("total_confirmed_revenue")
    )
    .orderBy("e.city")
)

revenue_df.show()
```
